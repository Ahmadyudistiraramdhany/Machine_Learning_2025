Data dan Persiapan

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Data Cuaca (Sample 10 Baris)
data_cuaca = {
    'Temperature': [14.0, 39.0, 30.0, 38.0, 27.0, 32.0, -2.0, 3.0, 3.0, 28.0],
    'Humidity': [73, 96, 64, 83, 74, 55, 97, 85, 83, 74],
    'Wind Speed': [9.5, 8.5, 7.0, 1.5, 17.0, 3.5, 8.0, 6.0, 6.0, 8.5],
    'Precipitation (%)': [82.0, 71.0, 16.0, 82.0, 66.0, 26.0, 86.0, 96.0, 66.0, 107.0],
    'Cloud Cover': ['partly cloudy', 'partly cloudy', 'clear', 'clear', 'overcast', 'overcast', 'overcast', 'partly cloudy', 'partly cloudy', 'clear'],
    'Atmospheric Pressure': [1010.82, 1011.43, 1018.72, 1026.25, 990.67, 1010.03, 990.87, 984.46, 999.44, 1012.13],
    'UV Index': [2, 7, 5, 7, 1, 2, 1, 1, 0, 8],
    'Season': ['Winter', 'Spring', 'Spring', 'Winter', 'Winter', 'Summer', 'Winter', 'Winter', 'Winter', 'Winter'],
    'Visibility (km)': [3.5, 10.0, 5.5, 1.0, 2.5, 5.0, 4.0, 3.5, 1.0, 7.5],
    'Location': ['inland', 'inland', 'mountain', 'coastal', 'mountain', 'inland', 'inland', 'inland', 'mountain', 'coastal'],
    'Weather Type': ['Rainy', 'Cloudy', 'Sunny', 'Sunny', 'Rainy', 'Cloudy', 'Snowy', 'Snowy', 'Snowy', 'Sunny']
}
df_cuaca = pd.DataFrame(data_cuaca)

# Pemrosesan Data (Encoding dan Scaling)
# Pisahkan Fitur (X) dan Target (y)
X = df_cuaca.drop('Weather Type', axis=1)
y = df_cuaca['Weather Type']

# One-Hot Encoding untuk fitur kategori (teks)
X_encoded = pd.get_dummies(X, columns=['Cloud Cover', 'Season', 'Location'], drop_first=True)

# Scaling Fitur Numerik (Penting untuk KNN agar jarak antar fitur adil)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)
X_scaled_df = pd.DataFrame(X_scaled, columns=X_encoded.columns)

print("Data Setelah Pemrosesan (5 Baris Pertama):")
print(X_scaled_df.head())
```

Implementasi KNN

```python
# Karena data sangat kecil, kita gunakan Cross-Validation, tetapi di sini hanya contoh training
k_cuaca = 3

# Latih Model
knn_model = KNeighborsClassifier(n_neighbors=k_cuaca)
knn_model.fit(X_scaled, y)

# Contoh Titik Uji: Kita ambil data Baris ke-5 (Aktual: Cloudy)
idx_uji = 5
X_test_point = X_scaled_df.iloc[[idx_uji]]

# Prediksi
prediction_cuaca = knn_model.predict(X_test_point)
probability_cuaca = knn_model.predict_proba(X_test_point)

print(f"Prediksi Cuaca Menggunakan KNN (k={k_cuaca})")
print(f"Data Uji: Baris ke-{idx_uji}")
print(f"Prediksi Jenis Cuaca: {prediction_cuaca[0]} (Aktual: {y[idx_uji]})")

# Menampilkan Probabilitas
classes = knn_model.classes_
prob_df = pd.DataFrame(probability_cuaca, columns=classes)
print("\nProbabilitas Klasifikasi:")
print(prob_df.T.to_string(header=False, float_format="%.2f"))
```
