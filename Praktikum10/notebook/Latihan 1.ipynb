Persiapan Data

```python
# Import library
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Data Set (Training) dari gambar Latihan 1
data = {
    'Temperatur Udara (°C)': [10, 25, 15, 20, 18, 20, 22, 24],
    'Kecepatan Angin (km/jam)': [0, 0, 5, 3, 7, 10, 5, 6],
    'Klasifikasi atau Persepsi Marry': ['Dingin', 'Panas', 'Dingin', 'Panas', 'Dingin', 'Dingin', 'Panas', 'Panas']
}
df = pd.DataFrame(data)

# Pisahkan fitur (X) dan target (y)
X = df[['Temperatur Udara (°C)', 'Kecepatan Angin (km/jam)']]
y = df['Klasifikasi atau Persepsi Marry']

# Titik data yang akan diuji (Data Test)
# Temperatur 16°C dan Kecepatan Angin 3 km/jam
X_test_point = pd.DataFrame([[16, 3]], columns=['Temperatur Udara (°C)', 'Kecepatan Angin (km/jam)'])
print("DataFrame Data Training:")
print(df)
print("\nDataFrame Titik Uji:")
print(X_test_point)
```

Mencari Nilai $k$ Terbaik (Cross-Validation)

```python
print("\nMenentukan Nilai k Terbaik Berdasarkan Akurasi")

best_k = 0
best_accuracy = 0
k_values = [1, 3, 5, 7] # Nilai k harus ganjil untuk menghindari seri dalam klasifikasi biner

for k in k_values:
    # Latih model KNN
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X, y)
    y_pred_train = knn.predict(X)
    accuracy = accuracy_score(y, y_pred_train)

    print(f"K={k}: Akurasi pada Data Training = {accuracy:.2f}")

    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_k = k

print(f"\nNilai k terbaik berdasarkan akurasi tertinggi pada data training adalah: k={best_k} (Akurasi: {best_accuracy:.2f})")
```

Klasifikasi Data Test

```python
# Gunakan k terbaik untuk klasifikasi (asumsi k=3 sebagai contoh umum, atau k terbaik yang didapat)
k_final = best_k # Atau k=3 jika ingin nilai k ganjil yang umum

knn_final = KNeighborsClassifier(n_neighbors=k_final)
knn_final.fit(X, y)

# Prediksi untuk Data Test (16°C, 3 km/jam)
persepsi_marry = knn_final.predict(X_test_point)

print(f"\nHasil Klasifikasi Data Test (k={k_final})")
print(f"Jika suhu 16°C dan kecepatan angin 3 km/jam, persepsi Marry adalah: {persepsi_marry[0]}")
```
