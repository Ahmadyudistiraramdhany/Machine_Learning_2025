Data dan Confusion Matrix

```python
# Pustaka untuk visualisasi data statistik
import pandas as pd
import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
import matplotlib.pyplot as plt
import seaborn as sns

# Data
# Label 'Lulus' dianalogikan sebagai kelas Positif (1) dan 'Tidak Lulus' sebagai Negatif (0)
actual_labels = ['Lulus', 'Lulus', 'Lulus', 'Lulus', 'Lulus',
                 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus']
predicted_labels = ['Lulus', 'Lulus', 'Lulus', 'Lulus', 'Tidak Lulus',
                    'Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus']

# Konversi label string menjadi numerik (1 = Lulus, 0 = Tidak Lulus)
y_true = np.array([1 if x == 'Lulus' else 0 for x in actual_labels])
y_pred = np.array([1 if x == 'Lulus' else 0 for x in predicted_labels])

# Confusion Matrix (Perhitungan)
# labels=[0, 1] memastikan urutan: 0=Tidak Lulus (Negatif), 1=Lulus (Positif)
cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
TN, FP, FN, TP = cm.ravel()

print("1. Confusion Matrix (Data Numerik)")
print(f"TP={TP}, TN={TN}, FP={FP}, FN={FN}")
print("Matriks:\n", cm)


# Visualisasi Confusion Matrix dengan Matplotlib dan Seaborn
# Nama label untuk sumbu
labels = ['Tidak Lulus', 'Lulus']

plt.figure(figsize=(6, 5))
# Membuat heatmap dari Confusion Matrix
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=labels,
    yticklabels=labels,
    cbar=False
)

# Menambahkan label dan judul
plt.title('Confusion Matrix Kelulusan')
plt.ylabel('Aktual (Sebenarnya)')
plt.xlabel('Prediksi (Hasil Model)')
plt.show()


```

Menghitung Persentase Matrix

```python
# 4. Perhitungan Metrik Persentase
print("\n2. Perhitungan Metrik Persentase")

# Accuracy: (TP + TN) / Total
accuracy = accuracy_score(y_true, y_pred)
print(f"Akurasi (Accuracy) = ({TP} + {TN}) / {len(y_true)} = {accuracy:.4f} ({accuracy*100:.2f}%)")

# Precision: TP / (TP + FP)
precision = precision_score(y_true, y_pred, pos_label=1)
print(f"Presisi (Precision) = {TP} / ({TP} + {FP}) = {precision:.4f} ({precision*100:.2f}%)")

# Recall: TP / (TP + FN)
recall = recall_score(y_true, y_pred, pos_label=1)
print(f"Recall (Sensitivitas) = {TP} / ({TP} + {FN}) = {recall:.4f} ({recall*100:.2f}%)")
```
