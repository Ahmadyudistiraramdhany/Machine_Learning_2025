```python
# memanggil data set lewat gdrive
path = "/content/drive/MyDrive/Machine_Learning/Latihan/data"
```

```python
#Membaca data csv
import pandas as pd
df = pd.read_csv(path + "/dataset_satelit.csv")
df.head()
```

```python
# Menampilkan satatistik deskriptif
df.describe()
```

```python
from sklearn.model_selection import train_test_split
x_columns =["b1", "Sigma_VV", "Sigma_VH", "plia", "lia", "iafe", "gamma0_vv", "gamma0_vh", "beta0_vv", "beta0_vh"]
y_column = ["P"]

X = df[x_columns]
y = df[y_column]

x_train, x_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)

print(f"Bentuk x_train: {x_train.shape}")
print(f"Bentuk x_test: {x_test.shape}")
X.head()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# evaluasi model
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

y_pred = model.predict(x_test)
r2 = r2_score(y_test, y_pred)


print("Koefisien (Fitur):", model.coef_)
print("Intercept:"         , model.intercept_)
print("R2 (test):"         , r2)
print("MAE (Unit P):"        , mean_absolute_error(y_test, y_pred))
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
print("RMSE (Unit P):"       , rmse)
```

```python
import numpy as np
import pandas as pd

# Prediksi P dengan model
y_pred_test = model.predict(x_test)

# Ambil nama fitur pertama yang digunakan
first_feature_name = x_test.columns[0]

# Buat DataFrame hasil dengan kolom fitur pertama, P aktual, dan P prediksi
hasil = pd.DataFrame({
    first_feature_name: x_test[first_feature_name].to_numpy(),
    "P aktual (unit)": y_test.to_numpy().ravel(), # PERBAIKAN: Tambahkan .ravel() di sini
    "P prediksi (unit)": y_pred_test.ravel(), # PERBAIKAN: Tambahkan .ravel() di sini
})

# 1. Hitung selisih error
hasil["selisih error (unit P)"] = hasil["P prediksi (unit)"] - hasil["P aktual (unit)"]

# 2. Hitung akurasi per baris (%)
# Ganti P aktual 0 dengan NaN agar tidak error pembagian
denom = hasil["P aktual (unit)"].replace(0, np.nan)
hasil["Akurasi (%)"] = (1 - (hasil["selisih error (unit P)"].abs() / denom)).clip(lower=0, upper=1) * 100

# Tampilkan hasil
hasil
```
